## 1: Import Required Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

```python
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

## 4: Dataset Information

```python
print("Shape of Dataset:", df.shape)

print("\nDataset Information:")
df.info()
```

## 5: Identify Features

```python
# Identifying numerical features
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Identifying categorical features
categorical_features = df.select_dtypes(include=['object']).columns.tolist()

# Target variable
target_variable = "Churn"

# Remove target variable from categorical features
if target_variable in categorical_features:
    categorical_features.remove(target_variable)

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nTarget Variable:")
print(target_variable)
```

### Observation

The dataset contains both numerical and categorical features. The target variable is **Churn**, which indicates whether a customer has left the telecommunications company. The identified features will be used in the preprocessing and model development stages.

## 6: Data Preprocessing

### 6.1 Check for Missing Values

```python
# Checking for missing values

df.isnull().sum()
```

```python
# Replacing blank spaces with NaN

df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)
```

```python
# Checking missing values again

df.isnull().sum()
```

### Observation

The dataset contained blank values in the **TotalCharges** column. These blank values were converted to missing values (NaN) so that they could be handled appropriately before training the model.

### 6.2 Handle Missing Values

```python
# Converting TotalCharges to numeric

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
```

```python
# Filling missing values with the median

df["TotalCharges"].fillna(df["TotalCharges"].median(), inplace=True)
```

```python
# Verifying that there are no missing values

df.isnull().sum()
```

### Observation

The missing values in the **TotalCharges** column were replaced with the median value. After preprocessing, the dataset contains no missing values.

### 6.3 Encode Categorical Variables

```python
# Creating a copy of the dataset

df_encoded = df.copy()
```

```python
# Initializing Label Encoder

label_encoder = LabelEncoder()
```

```python
# Encoding all categorical columns except customerID

categorical_columns = df_encoded.select_dtypes(include=["object"]).columns.tolist()

categorical_columns.remove("customerID")

for column in categorical_columns:
    df_encoded[column] = label_encoder.fit_transform(df_encoded[column])

df_encoded.head()
```

### Observation

All categorical variables except **customerID** were converted into numerical values using Label Encoding. The **customerID** column was excluded because it is only a unique identifier and does not contribute to predicting customer churn.

### 6.4 Split the Dataset

```python
# Defining features and target variable

X = df_encoded.drop(["customerID", "Churn"], axis=1)
y = df_encoded["Churn"]
```

```python
# Spliting the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)

print("Training Target  :", y_train.shape)
print("Testing Target   :", y_test.shape)
```

### Observation

The dataset was successfully divided into **80% training data** and **20% testing data**. The training data will be used to train the Logistic Regression model, while the testing data will be used to evaluate its performance.

## 7: Model Development

### 7.1 Build the Logistic Regression Model

```python
# Creating the Logistic Regression model

model = LogisticRegression(max_iter=1000)
```

```python
# Training the Logistic Regression model

model.fit(X_train, y_train)
```

### Observation

The Logistic Regression model was successfully trained using the training dataset. During training, the model learned the relationship between customer attributes and the likelihood of customer churn.

### 7.2 Predict Customer Churn

```python
# Predicting customer churn for the testing dataset

y_pred = model.predict(X_test)
```

```python
# Comparing Actual vs Predicted values

comparison = pd.DataFrame({
    "Actual Churn": y_test.values,
    "Predicted Churn": y_pred
})

comparison.head(10)
```

```python
comparison.head(10).style.format({
    "Actual Churn": "{:.0f}",
    "Predicted Churn": "{:.0f}"
})
```

### Observation

The trained Logistic Regression model successfully predicted customer churn for the testing dataset. A comparison between the actual and predicted values shows that the model is able to classify most customer records correctly, although some misclassifications are expected in any predictive model.
